```python
import pandas as pd
```

1) read_csv

```python
auto_df = pd.read_csv(
    "../data/auto.csv",
    sep=','
)
auto_df = auto_df.set_index("ID")
print(auto_df)
```

2) count

```python
print(auto_df.count())
```

3) убрать дупликаты

```python
auto_df = auto_df.drop_duplicates(subset=["CarNumber", "Make_n_model", "Fines"], keep="last")
print(auto_df.count())
```

4. отсутствующие значения

```python
def check_mis():#функция чтобы посмотреть скока отсутствующих значений в колонках
    for i in range(0,auto_df.shape[1]):
        col_name = auto_df.columns[i]
        missing_num = auto_df[auto_df[col_name].isna()].shape[0]
        print(f"{col_name}: {missing_num}")
```

```python
# for i in range(0,auto_df.shape[1]):
#     col_name = auto_df.columns[i]
#     missing_num = auto_df[auto_df[col_name].isna()].shape[0]
#     print(f"{col_name}: {missing_num}")
#     if (missing_num >= 500):
#         auto_df = auto_df.drop(columns=[col_name])
# auto_df.shape
```

```python
check_mis()
print(auto_df.shape)

#tresh=n не удаляет с >= n непропусоков,     
auto_df = auto_df.dropna(axis=1, thresh=auto_df.shape[0] - 499)

check_mis()
print(auto_df.shape)
```

```python
#ffil() - заполняет предыдущим значением, если нужно первое значение заполнить bfill
auto_df["Refund"] = auto_df["Refund"].ffill()
check_mis()
```

```python
mean = auto_df["Fines"].mean( skipna=True)
print(mean)
auto_df["Fines"] = auto_df["Fines"].fillna(mean)
check_mis()
```

5. Делим и сохраняем в Джейсон Вурхис

```python
split_df = auto_df["Make_n_model"].apply(lambda x: pd.Series(x.split()))
auto_df["Make"] = split_df[0]
auto_df["Model"] = split_df[1]
auto_df = auto_df.drop(columns="Make_n_model")
auto_df.to_json("../data/auto.json")
auto_df.head(2)
```
