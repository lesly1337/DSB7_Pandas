1) read_csv

```python
df = pd.read_csv(
    "../data/feed-views.log",
    sep=r"\s{2,}|\t",
    names=["datetime", "user"],
    engine="python",
    skiprows=[2,3],
    skipfooter=2 
)
df = df.set_index("datetime")
```

```python
df.count()
```

```python
print(df.head())
```

```python
print(df.tail())
```

2) datetime в date_time

```python
df = df.rename_axis("date_time")
print(df.head())
```

3) to_csv

```python
df.to_csv(
    "../data/feed-views-semicolon.log",
    sep=";",  
    encoding="utf-8"
)
```
