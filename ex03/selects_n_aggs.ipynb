1. Загружаем

```python
auto_df = pd.read_json("../data/auto.json")
auto_df = auto_df.set_index("CarNumber")
print(auto_df.head())
```

2. выбираем но с условием

```python
auto_df1 = auto_df[auto_df["Fines"] > 2100]
print(auto_df1)
```

```python
auto_df2 = auto_df[(auto_df["Fines"] > 2100) & (auto_df["Refund"] == 2)]
print(auto_df2)
```

```python
Models = ["Focus", "Corolla"]
auto_df3 = auto_df[auto_df["Model"].isin(Models)]
print(auto_df3)          
```

```python
Nums = ["Y7689C197RUS", "92928M178RUS", "7788KT197RUS", "H115YO163RUS", "X758HY197RUS"]
auto_df4 = auto_df.loc[Nums]
print(auto_df4)          
```

3. агрегация штрафов

```python
auto_df.groupby("Make")["Fines"].median()
```

```python
auto_df.groupby(["Make","Model"])["Fines"].median()
```

```python
auto_df.groupby(["Make","Model"])["Fines"].count()
```

```python
auto_df.groupby(["Make", "Model"])["Fines"].std()
```

4. агрегация машинных номеров

```python
bad_people = auto_df.groupby(auto_df.index)["Fines"].count().sort_values(ascending=False)
print(bad_people.head())
```

```python
worst = bad_people.index[0]
worst_row = auto_df.loc[worst]
print(worst_row)
```

```python
rich_people = auto_df.groupby(auto_df.index)["Fines"].sum().sort_values(ascending=False)
print(rich_people)
```

```python
smert_v_nichite = auto_df.loc[rich_people.index[0]]
print(smert_v_nichite)
```

Are there any car numbers connected to different models?

```python
model_per_number = auto_df.groupby(auto_df.index)["Model"].nunique()
print(model_per_number[model_per_number > 1])
```
