1. read csv

```python
fines_df = pd.read_csv("../data/fines.csv")
print(fines_df.head())
```

2. calc data (fines/refund*year)

```python
%%timeit
calc_list = [] 
for i in range(0,len(fines_df)):
    row = fines_df.iloc[i]
    calc_list.append(row["Fines"]/(row["Refund"]*row["Year"]))
fines_df["calc_data"] = calc_list
```

```python
%%timeit
calc_list = [] 
for index, row in fines_df.iterrows():
    calc_list.append(row["Fines"]/(row["Refund"]*row["Year"]))
fines_df["calc_data"] = calc_list
```

```python
%%timeit
fines_df["calc_data"] = fines_df.apply(lambda row: row["Fines"]/(row["Refund"]*row["Year"]), axis=1)
```

4. даункастинг

```python
fines_df.info(memory_usage='deep')
```

```python
optimized_df = fines_df.copy()
float64_t = optimized_df.select_dtypes(include=['float64']).columns
int64_t = optimized_df.select_dtypes(include=['int64']).columns
optimized_df[float64_t] = optimized_df[float64_t].astype('float32')
optimized_df["Refund"] = pd.to_numeric(optimized_df["Refund"], downcast='integer')
optimized_df["Year"] = pd.to_numeric(optimized_df["Year"], downcast='integer')
```

```python
optimized_df.info(memory_usage='deep')
```

5. категории

```python
obj_t = optimized_df.select_dtypes(include=['object']).columns
optimized_df[obj_t] = optimized_df[obj_t].astype('category')
```
